# **Random Forest Regressor**

---

<br>

## What is Random Forest Regressor?
- A supervised machine learning algorithm for **regression** tasks.
- Combines multiple **decision trees** to make more accurate predictions `(an ensemble method)`.

<br>

---

<br>

## How it Works:
1. **Creates Multiple Trees**:
   - Trains on random subsets of data (bagging) and features.
2. **Averaging**:
   - Predictions from all trees are averaged to get the final result.

<br>

---

<br>

## Advantages:
- **Accurate**: Handles complex datasets well and reduces overfitting.
- **Robust**: Works well with noisy data and outliers.
- **Feature Importance**: Identifies the most important input features.

<br>

## Disadvantages:
- **Slow**: Can be computationally expensive for large datasets.
- **Less Interpretable**: Difficult to understand compared to a single tree.

<br>

---

<br>

## Use Cases:
- **Price Prediction** (e.g., housing, stock prices)
- **Weather Forecasting**
- **Energy Consumption Estimation**

<br>

---

<br>

## Overview:
Random Forest Regressor is a robust, accurate algorithm that uses multiple decision trees and averaging to make precise continuous predictions.

<br>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
data = pd.read_csv('/content/51_Housing.csv')
data
```

```python
data.describe()
```

```python
data.isnull().sum()  # Find out Null Values - If it is 0 then No null values
```

```python
data.dropna(inplace=True)   # To drop null values from the table
```

```python
data.columns
```

```python
data = pd.get_dummies(data,columns = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning',
       'parking', 'prefarea', 'furnishingstatus'], drop_first=True, dtype = 'int')
```

```python
data
```

```python
X = data.drop(columns = 'price')
y = data['price']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state = 29)
```

```python
# Initialize the Random Forest Regressor

from sklearn.ensemble import RandomForestRegressor
rf_regressor = RandomForestRegressor(n_estimators=100, random_state=42)
```

```python
# Train the regressor
rf_regressor.fit(X_train, y_train)
```

```python
# Predict on the test set
y_pred = rf_regressor.predict(X_test)
```

```python
mse = mean_squared_error(y_test, y_pred)
print(mse)
```

```python
r2 = r2_score(y_test, y_pred)
print(r2)
```

```python
mae = mean_absolute_error(y_test, y_pred)
print(mae)
```
