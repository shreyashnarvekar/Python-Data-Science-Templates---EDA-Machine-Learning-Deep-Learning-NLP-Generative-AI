# **KNN : K-Nearest Neighbors**

---

<br>

**What is KNN?**
  - KNN is a simple algorithm for classification and regression.
  - It predicts a data point based on the `k` closest points to it.

<br>

---

<br>

**How KNN Works:**
  1. **Pick K:** Choose how many neighbors (k) to consider.
  2. **Calculate Distance:** Find the distance from the point to others (e.g., Euclidean).
  3. **Find Neighbors:** Select the `k` closest neighbors.
  4. **Make Prediction:**
     - **Classification:** Pick the most common class among the neighbors.
     - **Regression:** Average the values of the neighbors (for continuous output, like house prices).

<br>

---

<br>

**Advantages:**
  - Simple and easy to understand.
  - No assumptions about data distribution.
  - Effective for both classification and regression tasks.

<br>

**Disadvantages:**
  - Slow for large datasets.
  - Sensitive to irrelevant features.
  - Performance can degrade with high-dimensional data (curse of dimensionality).

<br>

---

<br>

**Key Factors:**
  - `k`: Number of neighbors.
  - `distance`: How we measure closeness (e.g., Euclidean).
  - `weights`: Whether closer neighbors count more (weighted distance).
  - `KNN Regressor`: For regression tasks, KNN predicts the average value of the `k` nearest neighbors.

<br>

---


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score, mean_squared_error
```

```python
df = pd.read_csv('/content/31_laptop_prices.csv')
```

```python
le = LabelEncoder()
df['Company'] = le.fit_transform(df['Company'])
df['Product'] = le.fit_transform(df['Product'])
df['TypeName'] = le.fit_transform(df['TypeName'])
df['OS'] = le.fit_transform(df['OS'])
df['Screen'] = le.fit_transform(df['Screen'])
df['Touchscreen'] = le.fit_transform(df['Touchscreen'])
df['IPSpanel'] = le.fit_transform(df['IPSpanel'])
df['RetinaDisplay'] = le.fit_transform(df['RetinaDisplay'])
df['CPU_company'] = le.fit_transform(df['CPU_company'])
df['CPU_model'] = le.fit_transform(df['CPU_model'])
df['PrimaryStorage'] = le.fit_transform(df['PrimaryStorage'])
df['SecondaryStorageType'] = le.fit_transform(df['SecondaryStorageType'])
df['PrimaryStorageType'] = le.fit_transform(df['PrimaryStorageType'])
df['GPU_company'] = le.fit_transform(df['GPU_company'])
df['GPU_model'] = le.fit_transform(df['GPU_model'])
```

```python
X = df.drop(["Price_euros", "Product"], axis = 1)
Y = df['Price_euros']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,Y,test_size = 0.2)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
k_values = list(range(1,11))
acc = []
mse = []
for k in k_values:
  knn = KNeighborsRegressor(n_neighbors=k, n_jobs=-1)  # KNN Regressor model
  knn.fit(X_train, y_train)  # Fit the model
  y_pred = knn.predict(X_test)  # Make predictions

  # Calculate Mean Squared Error for each k and store it
  error = mean_squared_error(y_test, y_pred)
  mse.append(error)

  # Calculate R2 score for each k and store it
  score = r2_score(y_test, y_pred)
  acc.append(score)
```

**MSE :** 2 has the lowest error, The model makes the least amount of mistake with this values of k compared to the others.

```python
plt.plot(k_values, mse, marker = 'o', )
```

**R2 Score :** 2 has the highest accuracy, The model performs the best with this values of k compared to the others.

```python
# Accuracy (R2 Score)
plt.plot(k_values, acc, marker = 'o', )
```
