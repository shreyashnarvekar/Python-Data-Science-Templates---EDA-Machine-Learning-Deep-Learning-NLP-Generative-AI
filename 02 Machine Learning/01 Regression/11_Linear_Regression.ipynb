# **LINEAR REGRESSION**

---

<br>

## **What is Linear Regression?**
- A statistical method to predict the target variable (dependent) using one or more features (independent).
- Assumes a **linear relationship** between the variables.

<br>

---

<br>

## **Equation**:

The general equation for Simple Linear Regression is:

$$ Y = \beta_0 + \beta_1 X + \epsilon $$

### Where:
- $Y$: Target variable (the value we are trying to predict).  
- $X$: Independent variable (the factor influencing $Y$).  
- $\beta_0$: Intercept (the constant value when $X = 0$).  
- $\beta_1$: Coefficient (the slope of the line, showing how $X$ affects $Y$).  
- $\epsilon$: Error term (the difference between predicted and actual values).


<br>

---

<br>

## **Key Points**:
1. **Goal**: Find the best-fitting straight line (regression line) for the data.  
2. **Prediction**: Uses the line equation to estimate the target variable for new data.

<br>

---

<br>

## **Advantages**:
- Simple to understand and implement.  
- Works well for small datasets with linear relationships.

<br>

---

<br>

## **Disadvantages**:
- Only models linear relationships.  
- Sensitive to outliers in the data.

<br>

---

<br>

## **Uses**:
- Predicting continuous outcomes (e.g., house prices, sales, etc.).

<br>

---


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
np.random.seed(34)
hours_studied = np.random.uniform(0,10,100)
exam_scores = 50 + 5 * hours_studied + np.random.normal(0,2,100)
```

```python
data = pd.DataFrame({'Hours_studies': hours_studied, 'Exam_score': exam_scores})
```

```python
data.head()
```

Split the data into train and test

```python
X = data[['Hours_studies']] # Independent Variable
Y = data[['Exam_score']]    # Target Variable
```

```python
X_train,X_test, Y_train,Y_test = train_test_split(X,Y, test_size = 0.2)
```

```python
model = LinearRegression()
model.fit(X_train,Y_train)
```

```python
Y_pred = model.predict(X_test)
```

```python
R2 = r2_score(Y_test,Y_pred)
```

```python
print(R2)
```

```python
plt.scatter(X_test, Y_test,color = 'b', label='Actual_data')
plt.plot(X_test,Y_pred, color = 'r', label = 'Regression line', linewidth = 2)
plt.xlabel('Hours studied')
plt.ylabel('Exam Scores')
plt.legend()
plt.show()
```

Finding Slope and Intercept

```python
slope = model.coef_[0]
intercept = model.intercept_

print(slope)
print(intercept)
```
