## **DBSCAN** - Density-Based Spatial Clustering of Applications with Noise

## What is DBSCAN?
Answer: DBSCAN is a clustering algorithm that groups together points that are closely packed, marking points in low-density regions as **outliers** (noise). It doesn't require you to specify the number of clusters beforehand.

<br>

---

## Key Parameters:
1. **`eps`**:
   - Maximum distance between two points for them to be considered neighbors.
   - Controls the size of the neighborhood around each point.
   
2. **`min_samples`**:
   - Minimum number of points required to form a dense region (a cluster).
   - Defines how many points must be in the `eps` neighborhood to form a cluster.

---

<br>

## Types of Points in DBSCAN:
1. **Core Points**:  
   Points that have at least `min_samples` within a distance of `eps`.
   
2. **Border Points**:  
   Points that are not core points but fall within the `eps` radius of a core point.

3. **Noise (Outliers)**:  
   Points that are neither core points nor within the `eps` radius of any core point.

<br>

---

<br>

## Advantages:
- Can find clusters of **any shape**.
- No need to specify the number of clusters in advance.
- Robust to noise and outliers.

<br>

## Disadvantages:
- Sensitive to the choice of **`eps`** and **`min_samples`**.
- Not well-suited for data with varying densities.

<br>

---

<br>

## When to Use DBSCAN:
- When the data has **clusters of varying shapes**.
- When you want to detect **outliers**.
- When the number of clusters is **not known beforehand**.

```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
X, y = make_moons(n_samples = 250, noise = 0.05)
```

```python
X
```

```python
plt.scatter(X[:,0], X[:,1])
```

```python
# Feature Scaling (Standard Scaling)
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

DBSCAN Alogrithm

```python
from sklearn.cluster import DBSCAN
```

`eps` is the maximum distance between two points for them to be considered neighbors and potentially belong to the same cluster. It controls the size of the neighborhood around each point.

```python
dbcan = DBSCAN(eps = 0.5)
```

```python
dbcan.fit(X_scaled)
```

```python
dbcan.labels_
```

```python
plt.scatter(X[:,0], X[:,1], c=dbcan.labels_)
```

```python
plt.scatter(X[:,0], X[:,1], c=y)
```
