# **KNN : K-Nearest Neighbors**

---

<br>

**What is KNN?**
  - KNN is a simple algorithm for classification (and regression).
  - It predicts a data point based on the `k` closest points to it.

<br>

---

<br>

**How KNN Works:**
  1. **Pick K:** Choose how many neighbors (k) to consider.
  2. **Calculate Distance:** Find the distance from the point to others (e.g., Euclidean).
  3. **Find Neighbors:** Select the `k` closest neighbors.
  4. **Make Prediction:**
     - **Classification:** Pick the most common class among the neighbors.
     - **Regression:** Average the values of the neighbors (for continuous output, like house prices).

<br>

---

<br>

**Advantages:**
  - Simple and easy to understand.
  - No assumptions about data distribution.
  - Works well for classification tasks, especially for small datasets.

<br>

**Disadvantages:**
  - Slow for large datasets.
  - Sensitive to irrelevant features.
  - Performance can degrade with high-dimensional data (curse of dimensionality).
  - Memory-intensive because the entire dataset is used for prediction.

<br>

---

<br>

**Key Factors:**
  - `k`: Number of neighbors.
  - `distance`: How we measure closeness (e.g., Euclidean).
  - `weights`: Whether closer neighbors count more (weighted distance).
  - `KNN Classifier`: For classification tasks, KNN assigns the most common class among the `k` nearest neighbors.

<br>

---


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
df = pd.read_csv('/content/31_iris.csv')
```

```python
df.head()
```

```python
X = df.drop('species', axis = 1)
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
```

### Key Points:
- **Loop:** We test the model for each `k` value in the range from 1 to 10.
- **Training and Prediction:** For each `k`, we train the model and predict the test data.
- **Storing Accuracy:** We store the accuracy for each `k` to compare and find the best value.


```python
k_values = list(range(1,11))
acc = []
for k in k_values:
  knn = KNeighborsClassifier(n_neighbors = k, n_jobs = -1)
  knn.fit(X_train, y_train)
  y_pred = knn.predict(X_test)
  accuracy = accuracy_score(y_test, y_pred)
  acc.append(accuracy)
```

```python
plt.plot(k_values, acc, marker = 'o', )
```
