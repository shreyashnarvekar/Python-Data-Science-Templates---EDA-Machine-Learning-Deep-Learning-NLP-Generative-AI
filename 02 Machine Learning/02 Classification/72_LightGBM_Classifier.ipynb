## 1. What is LightGBM?
- **LightGBM** (Light Gradient Boosting Machine) is an efficient and scalable gradient boosting library developed by Microsoft.
- It uses a **leaf-wise** tree growth algorithm that provides better accuracy and efficiency compared to traditional **level-wise** methods.
- LightGBM is widely used in supervised learning tasks such as classification and regression.

<br>

---

<br>

## 2. Key Features
- **Boosting Framework**: Builds an ensemble of weak learners (decision trees) to improve predictions.
- **Leaf-wise Tree Growth**: Unlike traditional algorithms, LightGBM grows trees leaf-wise, which can result in higher accuracy with fewer trees.
- **Faster Training**: Utilizes histogram-based methods for faster computation.
- **Handling Large Datasets**: LightGBM is optimized to handle large datasets with lower memory usage.
- **Cross-Platform Support**: Compatible with Python, R, C++, and other languages.

## 4. Advantages
- **High Speed**: Fast training, especially on large datasets, due to histogram-based algorithms.
- **Low Memory Usage**: Optimized for memory efficiency, allowing you to work with large datasets.
- **Handling Categorical Features**: Directly handles categorical features without the need for one-hot encoding.
- **Parallel and GPU Support**: Can leverage parallel computing and GPU acceleration for even faster performance.
- **Feature Importance**: Provides useful insights into the importance of different features in making predictions.

<br>

<br>

## ⭐ Things to Keep in Mind
- Ensure that **LightGBM** and **scikit-learn** are compatible.
- **Hyperparameter tuning** is crucial for **maximizing performance**.
- Be cautious of **overfitting** on smaller datasets; always use **cross-validation**.

````python
import pandas as pd
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import warnings
````

````python
# Suppress LightGBM specific warnings about splits
warnings.filterwarnings('ignore', category=UserWarning, message='No further splits with positive gain')

# Suppress sklearn future warning
warnings.filterwarnings('ignore', category=FutureWarning, message='.*_get_tags.*')

# Suppress specific FutureWarnings related to 'force_all_finite'
warnings.filterwarnings('ignore', category=FutureWarning, message="'force_all_finite'")
````

````python
data = datasets.load_breast_cancer()
X = data.data
y = data.target
````

````python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2)
````

````python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
````

````python
import lightgbm as lgb
````

> Hyperparameters for LightGBM Classifier

```python
model = lgb.LGBMClassifier(
    # Model Hyperparameters
    objective='binary',              # For binary classification (can also use 'multiclass' for multi-class classification)
    num_boost_round=100,             # Number of boosting rounds (trees)
    max_depth=6,                     # Maximum depth of each tree
    learning_rate=0.3,               # Step size at each iteration while moving toward a minimum of the loss function
    subsample=1.0,                   # Fraction of samples used for each tree to prevent overfitting
    colsample_bytree=1.0,            # Fraction of features used for each tree to prevent overfitting
    scale_pos_weight=1,              # Controls the balance of positive and negative weights for imbalanced classes
    
    # Regularization Parameters
    lambda_l1=0,                     # L1 regularization term on weights (Lasso)
    lambda_l2=1,                     # L2 regularization term on weights (Ridge)
    min_split_gain=0,                # Minimum loss reduction required to make a further partition (higher = more conservative)
    
    # Tree Growth Parameters
    min_child_samples=20,            # Minimum number of samples required to form a leaf node
    min_child_weight=1,              # Minimum sum of instance weight (hessian) in a child node
)
```

````python
# Initialize the LightBGM Classifier

model = lgb.LGBMClassifier(objective='binary',
                           verbose= -1,  # Suppress output (warnings and information),
                           random_state=42)
````

````python
# Train the dataset with LightBGM Model
model.fit(X_train, y_train)
````

````python
# Make predictions
y_pred = model.predict(X_test)
````

````python
# Evaluate the model
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

accuracy = accuracy_score(y_test, y_pred)
classification = classification_report(y_test, y_pred)
````

````python
print(f"Accuracy: {accuracy:.2f}")
````
