### 2. RegexpStemmer class
NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms. It basically takes a single regular expression and removes any prefix or suffix that matches the expression.

⬇ Let us see an example ⬇

````python
from nltk.stem import RegexpStemmer
````

````python
# The suffixes [ing, s, e, able] will be removed from any word that contains them.
reg_stemmer = RegexpStemmer('ing$|s$|able$', min = 4)
````

````python
reg_stemmer.stem('eating')
````

````python
reg_stemmer.stem('ingeating')
````

````python
# For prefixes, simply add a dollar sign '$' before the word
reg_stemmer = RegexpStemmer('$ing|$s|$able', min = 4)
````

````python
reg_stemmer.stem('ingeating')
````

````python
# For Prefixes and Suffixes, simply remove the dollar sign "$" from the word
reg_stemmer = RegexpStemmer('ing|s|able', min = 4)
````

````python
reg_stemmer.stem('ingeating')
````

> ## **WORDNET LEMMATIZER**

Lemmatization technique is like stemming. The output we will get after lemmatization is called `lemma`, which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses `morphy()` function to the `WordNet CorpusReader class` to `find a lemma`.

---

<br>

Use Case :
- **Q & A**
- **Chatbots**
- **Text Summarization**
- **Search Engines**
- **Topic Modeling**

<br>

---

⬇ Let us see it with an example ⬇



````python
import nltk
nltk.download('wordnet')
````

````python
from nltk.stem import WordNetLemmatizer
````

````python
lemmatizer = WordNetLemmatizer()
````

````python
# Default POS tag as 'n' for Noun
lemmatizer.lemmatize("going")
````

### POS (Part of Speech) Tag - `Hyperparameter` in Lemmatizer

```
Noun ---> n
Verb ---> v
adjective ---> a
adverb ---> r
```
<h3> Nouns: </h3>

1. **Dogs** → **Dog**
2. **Cars** → **Car**

<h3> Verbs: </h3>

1. **Running** → **Run**
2. **Better** (verb form) → **Bet**

<h3> Adjectives: </h3>

1. **Better** → **Good**
2. **Worse** → **Bad**

<h3> Adverbs: </h3>

1. **Quickly** → **Quick**
2. **Easily** → **Easy**





````python
# For Noun
lemmatizer.lemmatize("going", pos = 'n')
````

````python
# For Verb - (This is a correct for this word going)
lemmatizer.lemmatize("going", pos = 'v')
````

````python
# For Adjactive
lemmatizer.lemmatize("going", pos = 'a')
````

````python
# For Adverb
lemmatizer.lemmatize("going", pos = 'r')
````

````python
# Initializing list of words
words = ["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
````

````python
for word in words:
  print(word+" ----> "+lemmatizer.lemmatize(word))
````

````python
# POS 'v' works better in our case
for word in words:
  print(word+" ----> "+lemmatizer.lemmatize(word, pos = 'v'))
````

> ## WE ARE USING **TOKENIZATION**, **STOPWORDS REMOVAL** AND **STEMMING** OR **LEMMATIZATION** TOGETHER TO SIMPLIFY TEXT PROCESSING

- **Tokenization**: Splits text into smaller units like words or sentences (e.g., "The cats are running" → ["The", "cats", "are", "running"]).

- **Stopwords Removal**: Removes common, meaningless words like "is," "the," and "and" to focus on important terms.  

- **Stemming**: Reduces words to their root form to group similar words together (e.g., "running" → "run").

- **Lemmatization**: Converts words to their base form using a dictionary (e.g., "better" → "good").

5. **FW**: Foreign word (e.g., *bonjour*).  
6. **IN**: Preposition or subordinating conjunction (e.g., *in, because*).  
7. **JJ/JJR/JJS**: Adjectives (basic, comparative, superlative—e.g., *big, bigger, biggest*).  
8. **LS**: List marker (e.g., *1), a., b.*).  
9. **MD**: Modal verbs (e.g., *can, will*).

11. **NNP/NNPS**: Proper nouns (singular/plural—e.g., *India, Americans*).

14. **PRP/PRP\$**: Personal pronouns and possessive pronouns (e.g., *he, my*).  
15. **RB/RBR/RBS**: Adverbs (basic, comparative, superlative—e.g., *silently, better, best*).  
16. **RP**: Particle (e.g., *give up*).  
17. **TO**: The word *to* (e.g., *to go*).  
18. **UH**: Interjection (e.g., *wow*, *uh*).  
19. **VB, VBD, VBG, VBN, VBP, VBZ**: Verb forms (e.g., *take, took, taking, taken, takes*).  
20. **WDT, WP, WP\$, WRB**: Wh- words (*which, who, whose, where*).

## **COMPARISON :** **WORD2VEC** vs **BAG OF WORDS (BOW)** vs **TF-IDF**

| **Feature**               | **Word2Vec**                                                                                  | **Bag of Words (BoW)**                                                | **TF-IDF**                                                           |
|---------------------------|-----------------------------------------------------------------------------------------------|------------------------------------------------------------------------|------------------------------------------------------------------------|
| **Semantic Representation** | Captures semantic meaning and word relationships (e.g., `king - man + woman ≈ queen`).        | Does not capture semantic relationships, treats words as independent. | Does not capture semantic relationships, treats words as independent. |
| **Dimensionality**         | Produces dense, low-dimensional vectors (e.g., 100–300 dimensions).                          | Sparse, high-dimensional vectors proportional to vocabulary size.      | Sparse, high-dimensional vectors proportional to vocabulary size.      |
| **Contextual Awareness**   | Considers word context via CBOW or Skip-gram.                                                | Ignores word context, uses raw counts of words.                        | Ignores word context, uses frequency weights.                          |
| **Handles Synonymy**       | Maps similar words (e.g., "happy," "joyful") to similar vectors.                             | Treats synonyms as separate features.                                  | Treats synonyms as separate features.                                  |
| **Generalization**         | Pretrained embeddings can be used for transfer learning, reducing data requirements.         | Requires domain-specific training and cannot leverage pretrained models.| Requires domain-specific training and cannot leverage pretrained models.|
| **Data Sparsity**          | Handles data sparsity well with dense embeddings.                                            | Struggles with sparse data due to reliance on raw counts.              | Struggles with sparse data due to reliance on frequency weights.       |
| **Arithmetic Operations**  | Supports vector arithmetic (e.g., `king - man + woman ≈ queen`).                             | No support for arithmetic operations.                                  | No support for arithmetic operations.                                  |
| **Scalability**            | Efficient for large corpora using negative sampling.                                         | Computational cost increases with larger vocabularies.                 | Computational cost increases with larger vocabularies.                 |
| **Suitability for Tasks**  | Suitable for a wide range of NLP tasks (e.g., classification, similarity).                   | Limited to tasks relying on frequency-based features.                  | Limited to tasks relying on frequency-based features.                  |


> ## **WORD2VEC**

- By **Tomas Mikolov** and his colleagues at **Google** in 2013
---

<br>

### Overview
Word2Vec is a neural network-based technique to represent words as vectors (word embeddings), capturing semantic relationships. It uses two architectures:
- **CBOW (Continuous Bag of Words):** Predicts a target word from its context words.
- **Skip-gram:** Predicts context words given a target word.

### Key Concepts
1. **Word Embeddings:** Dense, low-dimensional vector representations of words.
2. **Training Objective:** Minimize the distance between related word vectors (semantic similarity).
3. **Cosine Similarity:** Measures similarity between word vectors.

<br>

---

<br>

### Architectures
- **CBOW:**
  - Faster and effective for small datasets.
  - Computes the probability of a word given its context.

- **Skip-gram:**
  - Works well with large datasets.
  - Captures semantic and syntactic relationships better.
  - Predicts context words for a given target word.

<br>

---

<br>

### Steps in Word2Vec
1. **Tokenization:** Split text into words or tokens.
2. **Vocabulary Creation:** Map words to unique indices.
3. **Training Word Vectors:** Use CBOW or Skip-gram models.
4. **Embedding Extraction:** Resulting vectors are used in downstream tasks.

---

### Hyperparameters
- **Window Size:** Context window for words (e.g., 2 for CBOW/Skip-gram).
- **Vector Size:** Dimensionality of word embeddings (e.g., 100, 300).
- **Negative Sampling:** Number of negative samples for efficient training.
- **Epochs:** Number of iterations over the training data.

---

### Use Cases
- Sentiment Analysis
- Named Entity Recognition (NER)
- Machine Translation
- Similarity Search

---

### Libraries Supporting Word2Vec
- **Gensim:** Python library to train and use Word2Vec models.
- **TensorFlow/PyTorch:** Custom implementation for more flexibility.

---

#### Example - **Using Gensim :**
```python
from gensim.models import Word2Vec

# Sample data
sentences = [["cat", "say", "meow"], ["dog", "say", "woof"]]

# Train model
model = Word2Vec(sentences, vector_size=100, window=5, min_count=1, workers=4)

# Access word vector
vector = model.wv['cat']

# Similarity
similarity = model.wv.similarity('cat', 'dog')
